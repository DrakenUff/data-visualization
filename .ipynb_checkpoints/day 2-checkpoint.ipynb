```python
#histogram to understand the distribution
for i in df.select_dtypes(include= 'number').columns:
     sns.histplot (data=df, x=i)
     plt.show()
```

```python
#boxplot to identify outliers
for i in df.select_dtypes(include= 'number').columns:
    sns.boxplot(data= df, x=i)
    plt.show()
```

```python
#scatter plot to understand the relationship

for i in ['Year', 'Adult Mortality', 'infant deaths',
       'Alcohol', 'percentage expenditure', 'Hepatitis B', 'Measles ', ' BMI ',
       'under-five deaths ', 'Polio', 'Total expenditure', 'Diphtheria ',
       ' HIV/AIDS', 'GDP', 'Population', ' thinness  1-19 years',
       ' thinness 5-9 years', 'Income composition of resources', 'Schooling']:
    sns.scatterplot(data= df, x=i, y= 'Life expectancy ')
    plt.show()
```

**MISSING VALUE TREATMENTS**

```python
df.isnull().sum()
```

```python
df.columns
```

```python
#khali spacema tesko columnko median le fill garim  
for i in [' BMI ', 'Polio', 'Income composition of resources']:
    df[i] = df[i].fillna(df[i].median())
```

```python
df[[' BMI ','Polio','Income composition of resources']].isnull().sum()
```
